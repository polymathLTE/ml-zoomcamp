```python
df_asia_features = df_asia[['vehicle_weight','model_year']]
```

```python
X = df_asia_features.head(7).values
XTX = np.dot(X.T, X)
```

```python
XTX
```

```python
# invert XTX
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = np.dot(np.dot(XTX_inv, X.T), y)
w.sum()

```
